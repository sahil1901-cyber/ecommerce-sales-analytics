```python
import pandas as pd
import numpy as np
df = pd.read_csv('../data/ecommerce_sales_dataset.csv')
#print("Shape:", df.shape)
#print(df.head())
#df.info()
#df.isnull().sum()
#print("duplicate rows:", df.duplicated().sum())
#print(df['payment_mode'].unique())
#print(df['city'].unique())
df = df.drop_duplicates()
print(df.shape)
df['city'] = df['city'].str.title()
print(df['city'].unique())
df['payment_mode'] = df['payment_mode'].replace('cod', 'COD')
df['payment_mode'] = df['payment_mode'].replace('COD ', 'COD')
df['payment_mode'] = df['payment_mode'].replace('Cash on Delivery', 'COD')
print(df['payment_mode'].unique())
df['payment_mode'] = df['payment_mode'].fillna('COD')
df['city'] = df['city'].fillna('Unknown')
df['seller_name'] = df['seller_name'].fillna('Unknown Seller')
df['discount_percent'] = df['discount_percent'].fillna(df['discount_percent'].mean())
print(df.isnull().sum())
df['price'] = df['price'].abs()
print(df['price'].min())
print("Shape:", df.shape)
print(df.isnull().sum())
print(df['payment_mode'].unique())
print(df['city'].nunique())
df.to_csv('../data/ecommerce_cleaned.csv', index=False)
print("Saved!")
```
